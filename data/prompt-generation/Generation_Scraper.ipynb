```python
from selenium.webdriver.common.action_chains import ActionChains
from selenium.webdriver.common.by import By
from selenium import webdriver


driver = webdriver.Chrome()
driver.get('http://coolors.co/generate')
actions = ActionChains(driver)
```

```python
def cycle(n = 10):
    palettes = []
    for i in range(n):
        actions.send_keys(' ')
        actions.perform()
        elems = driver.find_elements(By.CLASS_NAME, 'generator_color_hex')
        palette = [elem.text for elem in elems]
        palettes.append(palette)
    return palettes

```

```python
input("Exit out of the popup tutorial then press enter...")
n = int(input("Enter the number of color palettes you want: "))

palettes = cycle(n)
```

```python
print(palettes)
```

```python
#Convert palettes to images
from PIL import Image, ImageDraw, ImageColor

def convert_color(s):
    return ImageColor.getrgb('#' + s)

SAVE_PATH = './palette_files/'
IMAGE_SIZE = (750, 55)
BG_COLOR = (0,0,0,0)

RECTS = []
for j in range(5):
    RECTS.append(((150*j,0),(150*(j+1)-1,54)))

for i in range(1,len(palettes)+1):
    palette = palettes[i-1]
    img = Image.new(mode = 'RGBA', size = IMAGE_SIZE, color = BG_COLOR)
    img_drawer = ImageDraw.Draw(img)
    for j in range(5):
        img_drawer.rectangle(RECTS[j], convert_color(palette[j]))
    img.save(f'{SAVE_PATH}{i}.png')
```
